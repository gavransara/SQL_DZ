```python
# 16.Написати функцију count_lines(path: str) -> int која враћа број непразних редова у фајлу
# Input: 'data/raw/teams.csv'
# Output: 5
import csv

def count_lines(path):
    with open(path, 'r') as f:
        reader = csv.reader(f)
        count = 0
        for row in reader:
            if row: # ne moze is not None, jer csv.reader() za prazan red vraca [], [] je False, pa je ovo dovoljno
                count += 1
    return count

count_lines('./teams.csv')
```

```python
# 17.Написати функцију read_column(path: str, column: str) -> list која враћа све вредности једне колоне CSV фајла
# Input: 'data/raw/teams.csv', 'city'
# Output: ['Bela Palanka', 'Niš', 'Beograd, Rakovica', 'Kraljevo']
# Ако колона не постоји, вратити празну листу []

def read_column(path, column):
    with open(path, 'r') as f:
        teams = list(csv.DictReader(f))
    if teams == []:
        return []
    elif column not in teams[0]:
        return []
    else:
        lista = [t[column] for t in teams]
        return lista

read_column('./teams.csv', 'city')
```

```python
# 18.Написати функцију write_key_value(path: str, data: dict, headers: list) -> None која речник уписује у CSV фајл са две колоне
# Input: 'data/processed/summary.csv', {'rows': 10, 'valid': 6}, ['metric', 'value']
# Садржај фајла након позива:
# metric,value
# rows,10
# valid,6

# zaklavlja nisu nuzno kljucevi, mora csv.writer

def write_key_value(path, data, headers):
    datarow = [[k,v] for k,v in data.items()]
    #print(datarow)
    with open(path, 'w') as f:
        writer = csv.writer(f)
        writer.writerow(headers)
        writer.writerows(datarow)


write_key_value('../processed/summary.csv', {'rows': 10, 'valid': 6}, ['metric', 'value'])
```

```python
# 20. Написати функцију read_rows(path: str) -> list која враћа све редове CSV фајла без заглавља, као листу листи
# Користити csv.reader
# Input: 'data/raw/teams.csv'
# Output: [['FK Jedinstvo', 'Bela Palanka', '34'], ['FK Radnički', 'Niš', '41'],
# ['FK Rakovica', 'Beograd, Rakovica', '28'], ['FK Sloga', 'Kraljevo', '19']]

def read_rows(path):
    read_rows = []
    with open(path, 'r') as f:
        reader = csv.reader(f)
        header = next(reader)
        for r in reader:
            read_rows.append(r)
    return read_rows

read_rows('./teams.csv')
```

```python
# 21. Написати функцију total_points(path: str) -> int која враћа збир бодова свих тимова
# Користити csv.DictReader
# Input: 'data/raw/teams.csv'
# Output: 122

def total_points(path):
    points = 0 
    with open(path, 'r') as f:
        reader = list(csv.DictReader(f))
        for r in reader:
            points += int(r.get('points',0))
    return points

total_points('./teams.csv')
```

```python
# 22. Написати функцију write_rows(path: str, rows: list) -> None која листу листи уписује у CSV
# фајл
# Input: 'data/processed/people.csv', [['name', 'age'], ['Ana', 20], ['Marko', 22]]
# Садржај фајла након позива:
# name,age
# Ana,20
# Marko,22

def write_rows(path, rows):
    with open(path, 'w') as f:
        writer = csv.writer(f)
        for r in rows:
            writer.writerow(r)

write_rows('../processed/people.csv',[['name', 'age'], ['Ana', 20], ['Marko', 22]] )
```

```python
# 23. Написати функцију write_dicts(path: str, rows: list) -> None која листу речника уписује у
# CSV фајл
# Користити csv.DictWriter; заглавља су кључеви првог речника
# Input: 'data/processed/people.csv', [{'name': 'Ana', 'age': 20}, {'name': 'Marko', 'age': 22}]
#Садржај фајла након позива
# name,age
# Ana,20
# Marko,22

def write_dicts(path, rows):
    with open(path, 'w') as f:
        writer = csv.DictWriter(f, fieldnames = list(rows[0].keys()))
        writer.writeheader()
        writer.writerows(rows)

write_dicts('../processed/people1.csv',[{'name': 'Ana', 'age': 20}, {'name': 'Marko', 'age': 22}])
```

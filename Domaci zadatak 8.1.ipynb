```python
# 1. Napisati funkciju get_even_positions(data: tuple) -> tuple koja vraća elemente torke koji se nalaze na parnim indeksima (0, 2, 4...).
# Input: (10, 20, 30, 40, 50, 60)
# Output: (10, 30, 50)

def get_even_positions(data):
    n = len(data)
    result = []
    for i in range(0,n):
        if i % 2 == 0:
            result.append(data[i])
    return tuple(result)

get_even_positions((10, 20, 30, 40, 50, 60))
#get_even_positions(())
```

```python
# 2. Napisati funkciju find_exclusive_students(group1: set, group2:set) -> set koja vraća studente koji se nalaze u prvoj grupi, ali ne i u drugoj.
# Input: {'Ana','Marko','Jovan','Milica'},{'Marko','Milica','Petar'}
# Output: {'Ana','Jovan'}

def find_exclusive_students(group1, group2):
    return group1 - group2


find_exclusive_students({'Ana','Marko','Jovan','Milica'},{'Marko','Milica','Petar'})
#find_exclusive_students({'Ana','Marko','Jovan'},{'Marko','Jovan','Ana','Milos'})
```

```python
# 3. Napisati funkciju clean_tags(text: str) -> str koja obrađuje tekst sa tagovima.
# Potrebno je: 
#   Razdvojiti tagove koji su odvojeni karakterom |
#   Ukloniti početne i krajnje razmake svakog taga
#   Sva slova pretvoriti u mala slova
#   Spojiti tagove karakterom -

# Input:' Python | DATA | SQL ' 
# Output:'python-data-sql'

def clean_tags(text):
    niska = text.split('|') # lista stringova
    niska1 = []
    for n in niska:
        n = n.lower().strip()
        niska1.append(n)
    return '-'.join(niska1)

clean_tags(' Python | DATA | SQL ')
```

```python
# 4. Napisati funkciju count_products(csv_lines: list) -> dict koja analizira porudžbine iz CSV podataka i vraća ukupan broj naručenih proizvoda. Svaki red sadrži ime kupca, naziv proizvoda i količinu, razdvojene zarezom. Prvi red predstavlja zaglavlje i treba ga zanemariti. Ako se isti proizvod pojavljuje u više redova, sabrati njegove količine.
# Input: ['customer,product,quantity','Alice,laptop,2','Bob,mouse,3','Alice,laptop,1','Charlie,keyboard,2', 'Bob,mouse,1']
# Output: {'laptop': 3, 'mouse': 4, 'keyboard': 2}

def count_products(csv_lines):
    total_inventory = {}
    for l in csv_lines[1:]:
        item = l.split(',')[1]
        if item not in total_inventory:
            total_inventory[item] = int(l.split(',')[2])
        else:
            total_inventory[item] = int(l.split(',')[2]) + int(total_inventory.get(item, 0))
    return total_inventory

count_products(['customer,product,quantity','Alice,laptop,2','Bob,mouse,3','Alice,laptop,1','Charlie,keyboard,2', 'Bob,mouse,1'])

#count_products(['customer,product,quantity','Alice,laptop,2','Bob,mouse,3','Alice,laptop,1','Charlie,keyboard,2', 'Bob,mouse,1', 'Bob,laptop,2', 'Rachael,mouse,3', 'Peter,usb,2'])

#count_products([])
```
